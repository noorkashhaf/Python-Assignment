```python
import math

def get_circle_area(radius):
    return math.pi * (radius ** 2)

area = get_circle_area(10)
print(area)
```

```python
def calculator(a, b, operation):

    if operation == 'a':
        return a + b
    elif operation == 's':
        return a - b
    elif operation == 'm':
        return a * b
    elif operation == 'd':
        return a / b
    else:
        return "Invalid operation"

print(calculator(2, 5, 'd'))
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Run test
r = Rectangle(5, 10)
print(r.area())
```

```python
class Shape:

    def __init__(self, name, length):
        self.name = name
        self.length = length

    def area(self):
        return 0


class Square(Shape):

    def __init__(self, name, length):
        super().__init__(name, length)

    def area(self):
        return self.length * self.length

    def describe(self):
        print("This is a:", self.name)


s = Square('square', 5)

print("The area is:")
print(s.area())

print(s.describe())
```
